# Fitts' law: how long does pointing take?

Fitts' law {cite}`fitts1954information` predicts the time $MT$ to move to a target of width $W$ at distance $D$:

$$MT = a + b \log_2\left(\frac{D}{W} + 1\right)$$

The log term is the **index of difficulty** (ID, in bits). Here we simulate noisy participant data, fit $a$ and $b$, and use the model to compare designs. **Replace the simulation with your own measurements** for the exercise.

```python
import numpy as np
import matplotlib.pyplot as plt

rng = np.random.default_rng(0)
true_a, true_b = 0.15, 0.12   # seconds, seconds/bit (typical order of magnitude for a mouse)

D = rng.choice([64, 128, 256, 512, 1024], size=300)   # distance in pixels
W = rng.choice([8, 16, 32, 64], size=300)             # target width in pixels
ID = np.log2(D / W + 1)
MT = true_a + true_b * ID + rng.normal(0, 0.05, size=ID.size)
```

```python
b, a = np.polyfit(ID, MT, 1)
print(f"fitted a = {a:.3f} s, b = {b:.3f} s/bit, throughput ~ {1/b:.1f} bits/s")

plt.scatter(ID, MT, s=8, alpha=0.5, label="trials")
xs = np.linspace(ID.min(), ID.max(), 50)
plt.plot(xs, a + b * xs, "r", label="fit")
plt.xlabel("Index of difficulty (bits)"); plt.ylabel("Movement time (s)")
plt.legend(); plt.show()
```
